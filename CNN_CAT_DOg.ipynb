```python
import zipfile

zip_path ='archive.zip'
extract_path = '/content/cat_dog_data'


with zipfile.ZipFile(zip_path,'r') as zip_ref:
  zip_ref.extractall(extract_path)

print("Dataset extracted sucessfully")
```

```python
training_path = '/content/cat_dog_data/training_set/training_set'
testing_path = '/content/cat_dog_data/test_set/test_set'
```

```python
from keras.models import Sequential
from keras.layers import Conv2D
from keras.layers import MaxPooling2D
from keras.layers import Flatten
from keras.layers import Dense
```

```python
classifier = Sequential()

classifier.add(
    Conv2D(
        32,
        (3, 3),
        input_shape=(64, 64, 3),
        activation='relu'
    )
)

classifier.add(
    MaxPooling2D(
        pool_size=(2, 2)
    )
)

classifier.add(
    Conv2D(
        32,
        (3, 3),
        activation='relu'
    )
)

classifier.add(
    MaxPooling2D(
        pool_size=(2, 2)
    )
)

classifier.add(
    Flatten()
)

classifier.add(
    Dense(
        units=128,
        activation='relu'
    )
)

classifier.add(
    Dense(
        units=1,
        activation='sigmoid'
    )
)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(
    rescale=1./255,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)

test_datagen = ImageDataGenerator(
    rescale=1./255
)

training_set = train_datagen.flow_from_directory(
    training_path,
    target_size=(64, 64),
    batch_size=32,
    class_mode='binary'
)

test_set = test_datagen.flow_from_directory(
    testing_path,
    target_size=(64, 64),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)
```

```python
print(training_set.class_indices)
print(test_set.class_indices)

print(training_set.samples)
print(test_set.samples)
```

```python
classifier.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
classifier.fit(
    training_set,
    validation_data=test_set,
    epochs=10
)
```

```python
from google.colab import files
from tensorflow.keras.preprocessing import image
import numpy as np

# Upload a new cat/dog image
uploaded = files.upload()

# Get filename
filename = list(uploaded.keys())[0]

# Load and resize image
img = image.load_img(
    filename,
    target_size=(64, 64)
)

# Convert to array
img = image.img_to_array(img)

# Add batch dimension
img = np.expand_dims(img, axis=0)

# Normalize pixel values
img = img / 255.0

# Predict
prediction = classifier.predict(img)

print("Prediction value:", prediction[0][0])

# Classification
if prediction[0][0] > 0.5:
    print("Prediction: Dog")
else:
    print("Prediction: Cat")
```
